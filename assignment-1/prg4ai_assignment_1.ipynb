```python
## question  no 1

students_list = []

for i in range(5):
    student_name = input("Enter student's name:")
    roll_number = int(input("Enter Roll no. "))
    python_marks = int(input("Enter python marks:"))
    AI = int(input("Enter AI marks: "))
    Maths = int(input("Enter Maths marks: "))
```

```python
  student_info = {
        "Name": student_name,
        "Roll no": roll_number,
        "Python": python_marks,
        "Artificial Intelligence": AI,
        "Maths": Maths
    }
```

```python
 students_list.append(student_info)
```

```python

def result(marks1, marks2, marks3):
    total_marks = marks1 + marks2 + marks3
    percentage = total_marks / 3

    if percentage >= 80:
        grade = "Grade: A"
    elif percentage >= 70:
        grade = "Grade: B"
    elif percentage >= 60:
        grade = "Grade: C"
    elif percentage >= 50:
        grade = "Grade: D"
    else:
        grade = "Grade: F"

    return total_marks, percentage, grade
```

```python
for student in students_list:
    total, percentage, grade = result(
        student["Python"],
        student["Artificial Intelligence"],
        student["Maths"]
    )
```

```python
    student["Total marks"] = total
    student["Percentage"] = percentage
    student["Grade"] = grade
```

```python
def analyze(sentence):
    words = sentence.split()
    total_words = len(words)
    characters = len(sentence)
    longest = max(words, key=len)
    shortest = min(words, key=len)

    vowels = 0

    for ch in sentence.lower():
        if ch in "aeiou":
            vowels += 1

    unique_words = set(words)
    alphabetical = sorted(unique_words)

    result = f"""
Total words: {total_words}
Total characters: {characters}
Longest words: {longest}
Shortest words: {shortest}
Total vowels: {vowels}
Unique words: {unique_words}
Alphabetical: {alphabetical}
"""
       print(result)
```

```python
with open("words-analysis.txt", "w") as file:
    file.write(result)
    file.write(f"\nTotal words: {total_words}\n")
    file.write(f"Longest word: {longest}\n")
    file.write(f"Shortest word: {shortest}\n")
    file.write(f"Total vowels: {vowels}\n")
    file.write(f"Unique words: {sorted(unique_words)}\n")

print("Analysis saved in words-analysis.txt")
```

```python
Product = {}

for i in range(5):
    name = input("Enter product name")
    Price = int(input("Enter Product Price"))
    Quantity = float(input("Enter Product Quantity"))

    Product[name] = {
        "Price": Price,
        "Quantity": Quantity
    }
```

```python
def calculate_Bill(Product):
    Bill = 0

    for name in Product:
        Total = Product[name]["Price"] * Product[name]["Quantity"]

        print(name, "Total Price", Total)
        Bill = Bill + Total

    return Bill

Bill = calculate_Bill(Product)
print("Complete Bill:", Bill)

if Bill > 10000:
    discount = Bill * 10 / 100
else:
    discount = 0

final_Bill = Bill - discount

print("Discount:", discount)
print("Final Bill:", final_Bill)
```

```python
expensive = ""

for name in Product:
    if expensive == "":
        expensive = name
    elif Product[name]["Price"] > Product[expensive]["Price"]:
        expensive = name

print("Most Expensive Product:", expensive)
```

```python
with open("Bill.txt", "w") as file:
    file.write("Complete Bill:" + str(Bill))
    file.write("Discount:" + str(discount))
    file.write("Final Bill:" + str(final_Bill))
    file.write("Expensive:" + str(expensive))

print("Bill saved in Bill.txt")

file.close()
```

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b

def modulus(a, b):
    return a % b
```

```python
while True:
    a = float(input("Enter first number:"))
    b = float(input("Enter second number:"))

    op = input("Enter operation (+,-,*,/,%):")

    if op == "+":
        print("Result:", add(a, b))

    elif op == "-":
        print("Result:", subtract(a, b))

    elif op == "*":
        print("Result:", multiply(a, b))

    elif op == "/":
        if b != 0:
            print("Result:", divide(a, b))
        else:
            print("Cannot divide by zero")

    elif op == "%":
        if b != 0:
            print("Result:", modulus(a, b))
        else:
            print("Invalid operation")

    again = input("Do you want to calculate again? (Yes/No):")

    if again.lower() != "yes":
        break

print("Calculator closed")
```

```python
menu = {
    1: ("Burger", 500),
    2: ("Pizza", 800),
    3: ("Fries", 250),
    4: ("Drink", 150)
}

total = 0
bill = ""

while True:
    print("\n1. Burger - Rs. 500")
    print("2. Pizza - Rs. 800")
    print("3. Fries - Rs. 250")
    print("4. Drink - Rs. 150")
    print("5. Finish Order")

    choice = int(input("Select item: "))

    if choice == 5:
        break

    if choice in menu:
        quantity = int(input("Enter quantity: "))

        name, price = menu[choice]
        amount = price * quantity
        total = total + amount

        bill += f"{name} x {quantity} = Rs. {amount}\n"
        print(name, "added to order")

    else:
        print("Invalid choice")

print("\nYour Bill:")
print(bill)
print("Total Bill = Rs.", total)

with open("orders.txt", "w") as file:
    file.write(bill)
    file.write(f"Total Bill = Rs. {total}")

print("Bill saved in orders.txt")
```
